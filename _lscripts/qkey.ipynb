```python
from schubmult import *
from schubmult.abc import *
from IPython.display import display, Math
from sympy import *
QDG = QuantumDoubleGrothendieckRing(Sx.genset, ChangeOfVariables(z, y, lambda index, gset: 1 - 1/gset[index]), beta\
=-1)
def project(expr, n, ring=QDG):
    return ring.from_dict({k: v for k, v in expr.items() if k.inv == 0 or len(k) <= n})

display(Math(latex(project((QDG([2,3,4,1])*QDG([3,2,1])).expand(deep=False).simplify(),4))))
```
